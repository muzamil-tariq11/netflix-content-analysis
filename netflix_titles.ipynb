## 3. Data Cleaning

```python
df.columns
```

```python
df.isnull().sum()  # finding missing values
```

```python
df['date_added'] = df['date_added'].fillna(0)
df['rating'] = df['rating'].fillna(0)           # fill the missing values with 0
df['duration'] = df['duration'].fillna(0)
```

```python
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')         # fill the missing values with Unknown
df['country'] = df['country'].fillna("Unknown")
```

## 5. Visualization

```python
import matplotlib.pyplot as plt   # Visualization: Movies vs TV Shows

df['type'].value_counts().plot(kind='bar', color=['red', 'blue'])
plt.title('Movies vs TV Shows on Netflix')
plt.xlabel('Type')
plt.ylabel('Count')
plt.show()
```

```python
top_countries = df['country'].value_counts().head(10)
top_countries.plot(kind = 'bar' , color = 'green')
plt.title('Top 10 Countries by Netflix Content')
plt.xlabel('country')
plt.ylabel('Number of Titles')    
plt.xticks(rotation = 45)
plt.show()                # Visualization: Top 10 Content-Producing Countries
```

## 6. Key Insights

- Netflix has **6,131 Movies** and **2,676 TV Shows**, meaning movies make up the majority of content
- The **United States** produces the most content on Netflix, followed by other major countries
- The dataset covers content up to **2021**, with **592 titles** released in that year alone
- Missing values were found in `director` (2,634), `cast` (825), and `country` (831) — filled with "Unknown" for text columns and 0 for date/numeric columns
- The top 10 countries dominate a significant portion of Netflix's overall content catalog
